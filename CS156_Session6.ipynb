```python
import numpy as np
from sklearn import metrics
import matplotlib.pyplot as plt

NUM_DATAPOINTS = 1000

def sigmoid(x):
    return np.exp(x) / (1 + np.exp(x))

def generate_data(x):
    # true values
    a = 3  #beta_1
    b = -5  #beta_0
    prob = sigmoid(a*x+b)
    return np.random.uniform() < prob

x_data = np.random.uniform(high=3, size=NUM_DATAPOINTS)
y_data = [generate_data(x) for x in x_data]
```

```python
import jax
import jax.numpy as jnp

x = jnp.array(x_data)
y = jnp.array(y_data).astype(int)

def log_likelihood(beta_0, beta_1):
    p = jax.nn.sigmoid(beta_0 + beta_1 * x)
    return jnp.sum(y * jnp.log(p) + (1 - y) * jnp.log(1 - p))

gradient = jax.grad(log_likelihood, argnums=(0, 1))
grad_at_initial = gradient(3.0, -3.0)

print(grad_at_initial)
```
